```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("spotify_top_1000_tracks.csv")

student_name = "Manrique"
student_id = "TUPM-26-1137"

color1 = "#1f77b4"

print("Dataset loaded successfully!")
```

```python
# Additional Exercise 2: Song Release Trend

df['year'] = pd.to_datetime(
    df['release_date'],
    errors='coerce'
).dt.year

yearly_counts = df.groupby('year').size().sort_index()

cumulative_counts = yearly_counts.cumsum()

fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.bar(
    yearly_counts.index,
    yearly_counts.values,
    color=color1
)

ax1.set_xlabel("Year")
ax1.set_ylabel("Number of Songs")

ax2 = ax1.twinx()

ax2.plot(
    cumulative_counts.index,
    cumulative_counts.values,
    color='red',
    marker='o'
)

ax2.set_ylabel("Cumulative Songs")

plt.title(f"{student_name} ({student_id})\nSong Release Trend")
plt.tight_layout()
plt.show()
```

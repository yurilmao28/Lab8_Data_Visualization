```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("spotify_top_1000_tracks.csv")

# Student information
student_name = "YOUR NAME"
student_id = "YOUR STUDENT ID"

# Personalized color
color1 = "#1f77b4"

print("Dataset loaded successfully!")
```

```python
longest_tracks = df.nlargest(10, 'duration_min')

display(longest_tracks[
    ['track_name', 'artist', 'duration_min']
])

plt.figure(figsize=(12, 6))

plt.barh(
    longest_tracks['track_name'],
    longest_tracks['duration_min'],
    color=color1
)

plt.title(f"{student_name} ({student_id})\nTop 10 Longest Tracks")
plt.xlabel("Duration (minutes)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
